# 1. Introduction to Recursion

**Recursion** solves a problem by reducing it to a **smaller version of the same problem**.

A recursive function normally needs:

- **Base case** — stops the recursion.
- **Recursive case** — calls the function again with a smaller/simpler input.

> **Recursion = solve a smaller version of the same problem until a known simple case is reached.**


````python
def countdown(n):
    if n == 0:            # base case
        print("Go!")
        return

    print(n)
    countdown(n - 1)      # recursive case

countdown(5)
````

### Lecture questions

- What is the base case?
- What is the recursive case?
- Why does `n - 1` matter?

Each call moves closer to `0`, so the recursion eventually stops.


# 2. Training Example — Walking Steps

This example clearly shows the **call stack** and **unwinding**.


````python
def walk(steps):
    if steps == 0:
        print("Reached the base case")
        return

    print(f"Calling walk({steps - 1})")
    walk(steps - 1)
    print(f"You take step #{steps}")

walk(5)
````

# 3. Calculating a Power Recursively

For non-negative integer `n`:

$$x^n = x \times x^{n-1}$$

Base case:

$$x^0 = 1$$


````python
def power(x, n):
    if n == 0:
        return 1
    return x * power(x, n - 1)

print("5^3 =", power(5, 3))
print("2^6 =", power(2, 6))
````

# 4. Well-Defined Recursion

A recursive algorithm should:

1. Have a base case.
2. Make progress toward that base case.
3. Reach the base case in finite time.
4. Produce a consistent result for the same input.

A recursive call such as `countdown(n - 1)` is well-defined for a non-negative `n` because the problem becomes smaller each time.


## Infinite recursion

This is unsafe:

```python
def bad_countdown(n):
    print(n)
    bad_countdown(n + 1)
```

There is no stopping condition, and the input moves away from a base case.


````python
# Safe demonstration instead of true infinite recursion
def bad_countdown_demo(n, calls_left=5):
    print(f"bad_countdown({n})")

    if calls_left == 0:
        print("Stopped manually. Real infinite recursion would continue.")
        return

    bad_countdown_demo(n + 1, calls_left - 1)

bad_countdown_demo(1)
````

# 5. The Call Stack and Stack Overflow

Each active function call needs memory. Recursive calls may leave many functions waiting on the **call stack**.

If recursion becomes too deep, the system may run out of stack space. In Python, very deep recursion usually produces a `RecursionError`.


````python
import sys
print("Python recursion limit:", sys.getrecursionlimit())
````

# 6. Advantages of Recursion

Recursion is especially useful when a problem naturally breaks into smaller similar problems. It also supports **divide-and-conquer** thinking.

Common examples:

- factorial
- Fibonacci
- tree traversal
- Tower of Hanoi
- backtracking
- subset generation

Recursion is not automatically faster than iteration; often its main advantage is clarity.


# 7. Tail Recursion

In **tail recursion**, the recursive call is the final operation of the function.

An accumulator carries the partial result forward.


````python
def tail_sum(n, s=0):
    if n == 1:
        return n + s
    return tail_sum(n - 1, s + n)

print("1 + 2 + 3 + 4 =", tail_sum(4))
````

Trace:

```text
tail_sum(4, 0)
→ tail_sum(3, 4)
→ tail_sum(2, 7)
→ tail_sum(1, 9)
→ 10
```

The recursive call is the last action.

> Python does not guarantee tail-call optimisation, but tail recursion remains an important general concept.


# 8. Factorial with Recursion

$$0! = 1$$

$$n! = n(n-1)!$$

Example:

$$5! = 5\times4\times3\times2\times1 = 120$$


````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

for n in range(6):
    print(f"{n}! = {factorial(n)}")
````

# 9. Fibonacci Sequence

A common definition is:

$$F_0=0, \quad F_1=1$$

$$F_n = F_{n-1}+F_{n-2}$$

Sequence:

```text
0, 1, 1, 2, 3, 5, 8, 13, 21, ...
```


````python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print([fibonacci(i) for i in range(11)])
````

## Recursive Fibonacci call tree

For `F(5)`:

```text
                     F(5)
                   /      \
                F(4)      F(3)
               /   \      /   \
            F(3)  F(2)  F(2)  F(1)
```

The same subproblems are calculated repeatedly, which makes naive recursive Fibonacci inefficient for larger inputs.


````python
call_count = 0

def fibonacci_counted(n):
    global call_count
    call_count += 1
    if n <= 1:
        return n
    return fibonacci_counted(n - 1) + fibonacci_counted(n - 2)

for n in range(2, 11):
    call_count = 0
    value = fibonacci_counted(n)
    print(f"F({n}) = {value:>2}   calls = {call_count}")
````

## Iterative Fibonacci

The chapter also compares recursion with an iterative computation. The iterative version only needs the two most recent Fibonacci values.


````python
def fibonacci_iterative(n):
    if n == 0:
        return 0

    old_val = 0
    a = 1

    for _ in range(1, n):
        old_val, a = a, old_val + a

    return a

for n in range(11):
    print(f"F({n}) = {fibonacci_iterative(n)}")
````

# 11. Direct and Indirect Recursion

## Direct recursion
A function calls itself directly:

```text
A → A → A
```

## Indirect recursion
A function calls another function, which eventually calls the original function again:

```text
A → B → A → B
```


# 12. Squares with Direct Recursion

Use:

$$n^2 = (n-1)^2 + 2n - 1$$

with base case:

$$1^2=1$$


````python
def square_direct(n):
    if n == 1:
        return 1
    return square_direct(n - 1) + 2*n - 1

for n in range(1, 7):
    print(f"{n}^2 = {square_direct(n)}")
````

Call pattern:

```text
square(3) → square(2) → square(1)
```

This is **direct recursion**.


# 13. Squares with Indirect Recursion

The chapter's indirect example alternates between two functions:

```text
square → prod → square → prod ...
```


````python
def square_indirect(n):
    if n == 1:
        return 1
    return prod(n) + n

def prod(n):
    return square_indirect(n - 1) + (n - 1)

for n in range(1, 7):
    print(f"{n}^2 = {square_indirect(n)}")
````

| Type | Call pattern | Example |
|---|---|---|
| Direct | `A → A` | `square → square` |
| Indirect | `A → B → A` | `square → prod → square` |


# 14. Tower of Hanoi

The puzzle uses three pegs and disks of different sizes.

Rules:

1. Move one disk at a time.
2. Only move the top disk.
3. Never place a larger disk on a smaller disk.

Recursive strategy for `n` disks:

1. Move `n-1` disks from source to auxiliary.
2. Move the largest disk to destination.
3. Move the `n-1` disks from auxiliary to destination.


````python
def hanoi(n, source, auxiliary, destination):
    if n == 1:
        print(f"Move disk 1: {source} -> {destination}")
        return

    hanoi(n - 1, source, destination, auxiliary)
    print(f"Move disk {n}: {source} -> {destination}")
    hanoi(n - 1, auxiliary, source, destination)

hanoi(3, "A", "B", "C")
````

Minimum moves for `n` disks:

\[2^n - 1\]


````python
for n in range(1, 8):
    print(f"{n} disks -> {2**n - 1} moves")
````

# 15. Backtracking

**Backtracking** builds candidate solutions step by step.

At each step:

1. **Choose** an option.
2. **Explore** it.
3. If it fails, **undo** the choice.
4. Try another option.

> **Choose → Explore → Undo → Try another choice**


# 16. Generating All Subsets

For every element, we have two choices:

- include it,
- exclude it.

A set with `n` elements has:

$$2^n$$

subsets.


````python
def all_subsets(items):
    result = []

    def backtrack(index, current):
        if index == len(items):
            result.append(current.copy())
            return

        # Include
        current.append(items[index])
        backtrack(index + 1, current)

        # Undo
        current.pop()

        # Exclude
        backtrack(index + 1, current)

    backtrack(0, [])
    return result

subsets = all_subsets(["A", "B", "C"])
for s in subsets:
    print(s)
print("Number of subsets:", len(subsets))
````

# 17. Backtracking — Subset Sum

Find a subset of:

$$X = \{2,3,4,8\}$$

whose elements sum to:

$$6$$

Because all values are positive, if the running sum becomes greater than 6, that branch cannot succeed and can be **pruned**.


````python
def subset_sum(numbers, target):
    solution = None

    def backtrack(index, current, current_sum):
        nonlocal solution

        if current_sum == target:
            solution = current.copy()
            return True

        if current_sum > target:
            return False

        if index == len(numbers):
            return False

        # Include
        current.append(numbers[index])
        if backtrack(index + 1, current, current_sum + numbers[index]):
            return True
        current.pop()

        # Exclude
        if backtrack(index + 1, current, current_sum):
            return True

        return False

    backtrack(0, [], 0)
    return solution

X = [2, 3, 4, 8]
print("Solution:", subset_sum(X, 6))
````

A valid solution is:

$$\{2,4\}$$

because:

$$2+4=6$$


# 18. Visual Backtracking Trace

This version prints each choice and each backtrack.


````python
def subset_sum_trace(numbers, target):
    def backtrack(index, current, current_sum, depth=0):
        indent = "    " * depth
        print(f"{indent}current={current}, sum={current_sum}")

        if current_sum == target:
            print(f"{indent}SOLUTION FOUND: {current}")
            return True

        if current_sum > target:
            print(f"{indent}PRUNE: sum > {target}")
            return False

        if index == len(numbers):
            return False

        value = numbers[index]

        print(f"{indent}Include {value}")
        current.append(value)
        if backtrack(index + 1, current, current_sum + value, depth + 1):
            return True

        removed = current.pop()
        print(f"{indent}Backtrack: remove {removed}")

        print(f"{indent}Exclude {value}")
        return backtrack(index + 1, current, current_sum, depth + 1)

    return backtrack(0, [], 0)

subset_sum_trace([2, 3, 4, 8], 6)
````

## Exercise 7 — Subset Sum

Find a subset of `[1,3,5,7,9]` that sums to `12`.


````python
def all_subset_sums_answer(numbers, target):
    solutions = []
    def backtrack(index, current, current_sum):
        if current_sum == target:
            solutions.append(current.copy())
            return
        if current_sum > target or index == len(numbers):
            return
        current.append(numbers[index])
        backtrack(index + 1, current, current_sum + numbers[index])
        current.pop()
        backtrack(index + 1, current, current_sum)
    backtrack(0, [], 0)
    return solutions

print(all_subset_sums_answer([1, 3, 5, 7, 9], 12))
````

# 23. Chapter Summary

- **Recursion:** solve a problem using smaller versions of itself.
- **Base case:** stops recursion.
- **Recursive case:** reduces the problem and recurses.
- **Well-defined recursion:** reaches a base case in finite time.
- **Stack overflow:** too many active calls consume the call stack.
- **Tail recursion:** recursive call is the final operation.
- **Factorial:** classic single-branch recursion.
- **Fibonacci:** classic branching recursion with repeated work.
- **Direct recursion:** a function calls itself.
- **Indirect recursion:** functions call each other recursively.
- **Tower of Hanoi:** recursively moves `n-1`, then one disk, then `n-1` again.
- **Backtracking:** systematically explores choices and reverses failed choices.

> **Recursion reduces a problem; backtracking explores choices.**
